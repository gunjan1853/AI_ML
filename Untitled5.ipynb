```python
import zipfile

with zipfile.ZipFile("/content/microsoft-catsvsdogs-dataset.zip", "r") as zip_ref:
    zip_ref.extractall("/content/microsoft-catsvsdogs-dataset")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.layers import Dense,Flatten,Dropout
from keras.applications.vgg16 import VGG16
from tensorflow.keras import layers
```

```python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1),
    tf.keras.layers.RandomContrast(0.1),
])
```

```python
conv_base = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(150,150,3)
    )
```

```python
import os
from PIL import Image

DATA_DIR = "/content/microsoft-catsvsdogs-dataset/PetImages"

deleted = 0
kept = 0

for root, dirs, files in os.walk(DATA_DIR):

    for file in files:

        path = os.path.join(root, file)

        try:
            # Image open
            img = Image.open(path)

            # Actually read the image
            img.load()

            # Check channels
            channels = len(img.getbands())

            if channels not in [1, 3, 4]:
                print("Deleting invalid channels:", path)
                img.close()
                os.remove(path)
                deleted += 1
                continue

            # Convert RGBA / grayscale to RGB
            img = img.convert("RGB")

            # Save back as RGB
            img.save(path)

            img.close()

            kept += 1

        except Exception:
            # Empty / corrupted / unreadable image
            print("Deleting corrupted:", path)

            try:
                os.remove(path)
                deleted += 1
            except:
                pass


print("\n========== CLEANING COMPLETE ==========")
print("Valid images:", kept)
print("Deleted images:", deleted)
```

```python
model = Sequential([
    conv_base,
    Flatten(),
    Dense(256, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
conv_base.trainable=False
```

```python
model.summary()
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(150, 150),
    batch_size=32,
    label_mode="int"
)

validation_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(150, 150),
    batch_size=32,
    label_mode="int"
)
```

```python
normalization_layer = tf.keras.layers.Rescaling(1./255)

train_ds = train_ds.map(
    lambda x, y: (normalization_layer(x), y)
)

validation_ds = validation_ds.map(
    lambda x, y: (normalization_layer(x), y)
)
```

```python
from flax.nnx import optimizer
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)
```

```python
history = model.fit(
    train_ds,
    validation_data=validation_ds,
    epochs=15,
    callbacks=[early_stopping]
)
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['accuracy'], label='accuracy')
plt.plot(history.history['val_accuracy'], label = 'val_accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(loc='lower right')
plt.show()
```

```python
plt.plot(history.history['loss'], label='loss')
plt.plot(history.history['val_loss'], label = 'val_loss')
plt.xlabel('Epoch')
plt.ylabel('loss')
plt.legend(loc='lower right')
plt.show()
```

```python
model.evaluate(validation_ds)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np

y_true = []
y_pred = []

for images, labels in validation_ds:
    predictions = model.predict(images, verbose=0)

    y_true.extend(labels.numpy())
    y_pred.extend((predictions > 0.5).astype(int).flatten())

print(confusion_matrix(y_true, y_pred))

print(classification_report(
    y_true,
    y_pred,
    target_names=['Cat', 'Dog']
))
```
